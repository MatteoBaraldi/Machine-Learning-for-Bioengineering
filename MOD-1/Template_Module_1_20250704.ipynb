```python
In this exercise, you will evaluate a logistic regression model on an imbalanced binary classification
problem, using two different performance metrics: accuracy and AUC (Area Under the ROC Curve):
1. Data generation
The dataset will be provided in the code template. It consists of 1,000 samples and
20 numerical features, generated using make_classification. The data is strongly
imbalanced, with 95% of samples in class 0 (e.g., healthy) and 5% in class 1 (e.g.,
disease).
2. Model training
o Train a logistic regression model (LogisticRegression, from sklearn.linear_model)
using C = 1 as the regularization parameter.
o Use 5-fold cross-validation to evaluate model performance.
3. Evaluation
o Compute both accuracy and AUC on the test set of each fold.
o Report the mean and standard deviation of each metric across the folds.
4. Written discussion (in a Google Colab text cell):
o Does accuracy reflect how well the model identifies the minority class?
o In a clinical setting, which metric would be more appropriate and why?
```

```python
from sklearn.datasets import make_classification
# Step 1: Generate synthetic imbalanced dataset
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=15,
    n_redundant=5,
    n_clusters_per_class=2,
    weights=[0.95, 0.05],  # 95% class 0, 5% class 1
    random_state=42
)
```

Place below your code

```python
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.linear_model import LogisticRegression
import numpy as np
SEED = 42
n_folds = 5

clf = LogisticRegression(C=1, max_iter=1000)

# StratifiedKFold preserves the 95/5 class ratio in every fold
cv = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=SEED)

score = cross_validate(
    clf, X=X, y=y, cv=cv,
    scoring=['accuracy', 'roc_auc'],
    return_train_score=True,
)

acc = score['test_accuracy']
auc = score['test_roc_auc']

print("\nPer-fold Accuracy:", np.round(acc, 4))
print("Per-fold AUC:     ", np.round(auc, 4))
print(f"\nAccuracy: mean = {acc.mean():.4f}, std = {acc.std():.4f}")
print(f"AUC:      mean = {auc.mean():.4f}, std = {auc.std():.4f}")
```

```python
#1.because 95% belong to class 0,accuracy predicts majority class almost
#everytime, but has very low value for minority class: it weighs every samples
# equally so it is dominated by whichever class has more samples (it can be seen
# also from std which is nearly the same, while AUC change more )
# #2. AUC is more appropriate because is far less sensitive to class imbalance
# than accuracy so it's more honest, then evaluatees ranking across all
# threshold (can be changed based on costs of false positives/ false negatives)
```

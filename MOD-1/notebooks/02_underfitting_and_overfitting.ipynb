# 🎯 Underfitting and Overfitting – Interactive Demo

In this demo, we use polynomial regression to explore the concepts of **underfitting** and **overfitting**

- The true function we want to learn is a simple **quadratic** curve.
- We generate noisy observations of that curve.
- You can interactively adjust:
  - The **degree** of the polynomial model (its capacity)
  - The **noise level** added to the data

Use the sliders to observe how model complexity and noise affect both **training** and **test error**.

```python
# Install ipywidgets if not already available
!pip install ipywidgets --quiet

from ipywidgets import interact #interact is used to create interactive sliders
import numpy as np #=
import matplotlib.pyplot as plt #=
from sklearn.linear_model import LinearRegression #linear model used
from sklearn.preprocessing import PolynomialFeatures #transforms linear variables in polynomial ones
from sklearn.metrics import mean_squared_error #our performance metric
from sklearn.model_selection import train_test_split #function used to divide dataset in training set e test set

# Base data (same for all runs)
np.random.seed(0) #stops the random generator so its replicable
n_samples = 100 #define the number of samples
X = np.linspace(0, 1, n_samples).reshape(-1, 1) #first it creates 100 equispaced points between 0 and 1 and then transforms the vector in a 1 column matrix
# y_true = np.sin(2 * np.pi * X).ravel()
y_true = 4 * (X.ravel() - 0.5) ** 2 #ravel transforms X in a monodimensional vector and then its used to generate a perfect U parabola

# Interactive plot function (every time we use the slider)
def plot_polynomial_fit(degree=1, noise_std=0.1):
    # Generate noisy labels
    y = y_true + noise_std * np.random.randn(n_samples) #create real values adding noise

    # Train/test split
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) #divide in training and test set

    # Polynomial features
    poly = PolynomialFeatures(degree) #if degree = 3 this block takes x and returns [1,x,x^2,x^3]
    X_train_poly = poly.fit_transform(X_train) #fit implies the study of the data while tranform does the transformation
    X_test_poly = poly.transform(X_test) #just use the transformation based on the parameters obtained form training
    #if we were to use fit_ecc on the test we could risk data leakage, polluting the model with validation data

    # Train model
    model = LinearRegression() #initialize the model
    model.fit(X_train_poly, y_train) #trains the model on the training set y_train and finds the optimal parameters

    # Predictions and errors
    y_train_pred = model.predict(X_train_poly) #prediction on training set
    y_test_pred = model.predict(X_test_poly) #prediction on test set
    train_mse = mean_squared_error(y_train, y_train_pred) #mse on training
    test_mse = mean_squared_error(y_test, y_test_pred) #mse on test

    # Plotting
    X_plot = np.linspace(0, 1, 200).reshape(-1, 1) #create a dense line to have a continuous plot
    y_plot = model.predict(poly.transform(X_plot))

    plt.figure(figsize=(7, 5))
    plt.scatter(X_train, y_train, color='blue', label='Train data', s=20)
    plt.plot(X_plot, y_plot, color='red', label=f'Degree {degree}')
    plt.plot(X, y_true, '--', color='green', label='True function')
    plt.title(f"Train MSE = {train_mse:.4f} | Test MSE = {test_mse:.4f} | Noise std = {noise_std}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()

# Activate interactive widgets
interact(plot_polynomial_fit, degree=(1, 15), noise_std=(0.0, 1.0, 0.05))
```

# 📊 MSE vs Polynomial Degree

This plot shows how the **training** and **test Mean Squared Error (MSE)** change as we increase the degree of the polynomial model.

- We compare two scenarios:
  - Low noise (`noise_std = 0.15`)
  - High noise (`noise_std = 0.45`)
  
Key insights:
- Low-degree models may underfit (high error on both sets)
- High-degree models may overfit (low training error but high test error)
- Higher noise levels make overfitting occur earlier

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Base data
np.random.seed(0)
n_samples = 100
X = np.linspace(0, 1, n_samples).reshape(-1, 1)
#y_true = np.sin(2 * np.pi * X).ravel()
y_true = 4 * (X.ravel() - 0.5) ** 2
degrees = range(1, 16) #defines the degrees to test, we make them change automatically
noise_levels = [0.15, 0.45] #low noise scenario and high noise scenario

plt.figure(figsize=(10, 5)) #initializes the figure

for noise_std in noise_levels: #firs 0.15 then 0.45
    y = y_true + noise_std * np.random.randn(n_samples) #takes the noise from levels and generates the dataset
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) #divide in training and test set
    train_mse = [] #empty list at each noise change, used to store the training errors
    test_mse = [] #empty list at each noise change, used to store the test errors

    for degree in degrees: #for the present noise level it trains 15 different models (degrees)
        poly = PolynomialFeatures(degree) #initializes the transformation tool
        X_train_poly = poly.fit_transform(X_train) #study the training set
        X_test_poly = poly.transform(X_test) #just transformation

        model = LinearRegression() #initializes the model
        model.fit(X_train_poly, y_train) #training of the model, find the weights coefficients that minimize the distance from the training points

        y_train_pred = model.predict(X_train_poly) #asks for a prediction (stima) using training data
        y_test_pred = model.predict(X_test_poly) #idem but with test data (never seen during training)

        train_mse.append(mean_squared_error(y_train, y_train_pred)) #save the MSEs on the empty vectors
        test_mse.append(mean_squared_error(y_test, y_test_pred))

    # Plot for current noise level
    plt.plot(degrees, train_mse, linestyle='--', label=f'Train MSE (noise={noise_std})')
    plt.plot(degrees, test_mse, label=f'Test MSE (noise={noise_std})')

plt.xlabel('Polynomial Degree')
plt.ylabel('MSE')
plt.title('Training and Test MSE vs Model Capacity')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# 🎯 Underfitting and Overfitting – Interactive Demo

In this demo, we use polynomial regression to explore the concepts of **underfitting** and **overfitting**

- The true function we want to learn is a simple **quadratic** curve.
- We generate noisy observations of that curve.
- You can interactively adjust:
  - The **degree** of the polynomial model (its capacity)
  - The **noise level** added to the data

Use the sliders to observe how model complexity and noise affect both **training** and **test error**.

```python
# Install ipywidgets if not already available, needed for interactivity
!pip install ipywidgets --quiet

from ipywidgets import interact
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures #from linear regression start then increase space by adding more degrees
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Base data (same for all runs)
np.random.seed(0) #need to get same random noise
n_samples = 100
X = np.linspace(0, 1, n_samples).reshape(-1, 1)  #reshape turn from 1D array into a 2D colum vector (what scikit-learn expect)
# y_true = np.sin(2 * np.pi * X).ravel()
y_true = 4 * (X.ravel() - 0.5) ** 2   #this is the ground truth --> what model doesn't see but is supposed to learn

# Interactive plot function
def plot_polynomial_fit(degree=1, noise_std=0.1):
    # Generate noisy labels --> simulate real world data
    y = y_true + noise_std * np.random.randn(n_samples)

    # Train/test split --> 70% for training test and 30% as test set; random state fixed for reproducibility
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

    # Polynomial features --> learn transformation from training data and apply it on test data
    poly = PolynomialFeatures(degree) #this is a vector shaped like this [1, x, x², ..., x^degree]
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)

    # Train model --> linear regression include also polynomial so fit produces also curve in original X space
    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    # Predictions and errors --> predictions on both training and test sets and calculate MSE
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)
    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)
    #IF both error high underfitting, if train low but test high OVERFITTING, if both low and similar GOOD FIT

    # Plotting --> just for visualization, smooth dense grid of 200 values
    X_plot = np.linspace(0, 1, 200).reshape(-1, 1)
    y_plot = model.predict(poly.transform(X_plot))

    plt.figure(figsize=(7, 5))
    plt.scatter(X_train, y_train, color='blue', label='Train data', s=20)
    plt.plot(X_plot, y_plot, color='red', label=f'Degree {degree}')
    plt.plot(X, y_true, '--', color='green', label='True function') #true, noise free parabola
    plt.title(f"Train MSE = {train_mse:.4f} | Test MSE = {test_mse:.4f} | Noise std = {noise_std}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()

# Activate interactive widgets --> slides for degree and noise; every time I run plot_polynomial-fit reruns automatically
#at low degrees curve too simple (underfitting); for too high degrees overfitting, especially with high noise
interact(plot_polynomial_fit, degree=(1, 15), noise_std=(0.0, 1.0, 0.05))
```

# 📊 MSE vs Polynomial Degree

This plot shows how the **training** and **test Mean Squared Error (MSE)** change as we increase the degree of the polynomial model.

- We compare two scenarios:
  - Low noise (`noise_std = 0.15`)
  - High noise (`noise_std = 0.45`)
  
Key insights:
- Low-degree models may underfit (high error on both sets)
- High-degree models may overfit (low training error but high test error)
- Higher noise levels make overfitting occur earlier

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Base data
np.random.seed(0)
n_samples = 100
X = np.linspace(0, 1, n_samples).reshape(-1, 1)
#y_true = np.sin(2 * np.pi * X).ravel()
y_true = 4 * (X.ravel() - 0.5) ** 2  #until here same as above
degrees = range(1, 16)
noise_levels = [0.15, 0.45] #instead of one at the time we test every degree and from 1 to 15 and two fixed noise levels

plt.figure(figsize=(10, 5)) #4 curves plot --> 2 for train MSE and 2 for test MSE

for noise_std in noise_levels: #outer loop for noise levels, splitting once for train/test set
    y = y_true + noise_std * np.random.randn(n_samples)
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    train_mse = []
    test_mse = []

    for degree in degrees: #loop for degrees--> transform features, fit model, predict and record MSE for the 2 lists
        poly = PolynomialFeatures(degree)
        X_train_poly = poly.fit_transform(X_train)
        X_test_poly = poly.transform(X_test)

        model = LinearRegression()
        model.fit(X_train_poly, y_train)

        y_train_pred = model.predict(X_train_poly)
        y_test_pred = model.predict(X_test_poly)

        train_mse.append(mean_squared_error(y_train, y_train_pred))
        test_mse.append(mean_squared_error(y_test, y_test_pred))

    # Plot for current noise level: dashed line for train MSE across degrees, solid line for test MSE across degrees
    plt.plot(degrees, train_mse, linestyle='--', label=f'Train MSE (noise={noise_std})')
    plt.plot(degrees, test_mse, label=f'Test MSE (noise={noise_std})')

plt.xlabel('Polynomial Degree')
plt.ylabel('MSE')
plt.title('Training and Test MSE vs Model Capacity')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#FINAL CONSIDERATIONS --> I GET 4 CURVES BECAUSE TWO MSE FOR TWO DIFFERENT NOISE LEVELS
#TRAIN MSE decreases almost monotonically as degree increases --> increasing capacity fit the data better
#TEST MSE decreases at first, reaches a minimum with enough capacity then increases at higher degrees (starts fitting noise instead of signal)
#high test noise (0.45) overfit at a lower degree (curve of error go up before), because noisier data has less "true signal" to justify high-capacity model
```

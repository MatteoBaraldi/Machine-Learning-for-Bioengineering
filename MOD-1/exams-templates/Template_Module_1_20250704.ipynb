```python
from sklearn.datasets import make_classification

# Step 1: Generate synthetic imbalanced dataset
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=15,
    n_redundant=5,
    n_clusters_per_class=2,
    weights=[0.95, 0.05],  # 95% class 0, 5% class 1
    random_state=42
)
```

Place below your code

```python
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.linear_model import LogisticRegression
import numpy as np

SEED = 42
n_folds = 5

clf = LogisticRegression(C=1, max_iter=1000)

# StratifiedKFold preserves the 95/5 class ratio in every fold
cv = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=SEED)

score = cross_validate(
    clf, X=X, y=y, cv=cv,
    scoring=['accuracy', 'roc_auc'],
    return_train_score=True,
)

acc = score['test_accuracy']
auc = score['test_roc_auc']

print("\nPer-fold Accuracy:", np.round(acc, 4))
print("Per-fold AUC:     ", np.round(auc, 4))

print(f"\nAccuracy: mean = {acc.mean():.4f}, std = {acc.std():.4f}")
print(f"AUC:      mean = {auc.mean():.4f}, std = {auc.std():.4f}")
```
